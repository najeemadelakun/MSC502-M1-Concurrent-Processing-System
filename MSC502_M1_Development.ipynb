```python

from pathlib import Path
import subprocess

REPO_URL = (
    "https://github.com/najeemadelakun/"
    "MSC502-M1-Concurrent-Processing-System.git"
)
PROJECT_DIR = Path("/content/MSC502-M1-Concurrent-Processing-System")

def run_git(command):
    result = subprocess.run(
        command,
        check=True,
        capture_output=True,
        text=True
    )
    return result.stdout.strip()

if PROJECT_DIR.exists():
    if not (PROJECT_DIR / ".git").exists():
        raise RuntimeError(
            "The project directory exists but is not a Git repository."
        )

    origin = run_git([
        "git", "-C", str(PROJECT_DIR),
        "remote", "get-url", "origin"
    ])

    if origin.rstrip("/") != REPO_URL.rstrip("/"):
        raise RuntimeError("Repository URL does not match the approved project.")

    run_git([
        "git", "-C", str(PROJECT_DIR),
        "pull", "--ff-only", "origin", "main"
    ])
else:
    run_git([
        "git", "clone", "--branch", "main",
        REPO_URL, str(PROJECT_DIR)
    ])

print("Repository successfully prepared.")
print("Project directory:", PROJECT_DIR)
print("Branch:", run_git([
    "git", "-C", str(PROJECT_DIR),
    "branch", "--show-current"
]))
print("\nRepository files:")
for item in sorted(PROJECT_DIR.iterdir()):
    print(item.name + ("/" if item.is_dir() else ""))

print("\nGit status:")
print(run_git([
    "git", "-C", str(PROJECT_DIR), "status", "--short"
]) or "Working tree clean.")
```
